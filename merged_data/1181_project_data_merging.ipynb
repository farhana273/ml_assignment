```python

# Read the complete "Data" sheet from the World Bank Excel file
# This contains economic indicators for all countries and years
data = pd.read_excel("WDIEXCEL.xlsx", sheet_name="Data")

# Select only the rows containing GDP growth
# NY.GDP.MKTP.KD.ZG is the World Bank code for GDP growth (annual %)
gdp_data = data[data["Indicator Code"] == "NY.GDP.MKTP.KD.ZG"].copy()

# Select the country information and GDP growth values
# We are only looking at the years 2020 to 2024 for now
gdp_data = gdp_data[[
    "Country Name",
    "Country Code",
    "2020",
    "2021",
    "2022",
    "2023",
    "2024"
]]

# Display the first 10 rows
print(gdp_data.head(10))

# Display how many rows we extracted
print("Number of rows:", len(gdp_data))
```

```python

# Read the "Country" sheet from the World Bank Excel file
# This sheet contains information about countries and regions
country_info = pd.read_excel(
    "WDIEXCEL.xlsx",
    sheet_name="Country"
)

# Display all column names
# This helps us identify which columns contain country information
print("Column names:")
print(country_info.columns.tolist())

# Display the first 5 rows
# This helps us understand how the country information is organised
print("\nFirst 5 rows:")
print(country_info.head())
```

```python

# Step 1: Select individual countries from the Country sheet
# We keep rows where the "Region" column is not empty
valid_countries = country_info[
    country_info["Region"].notna()
]

# Step 2: Get the country codes for individual countries
valid_country_codes = valid_countries["Country Code"]

# Step 3: Filter our GDP dataset
# Keep only rows whose country code is in our valid country list
gdp_countries = gdp_data[
    gdp_data["Country Code"].isin(valid_country_codes)
].copy()

# Step 4: Compare the number of rows before and after filtering
print("Before filtering:", len(gdp_data))
print("After filtering:", len(gdp_countries))

# Step 5: Display the first 10 rows of the cleaned dataset
print("\nGDP growth data for individual countries:")
display(gdp_countries.head(10))
```

```python

# Step 1: Select only inflation data from the World Bank dataset
# FP.CPI.TOTL.ZG is the code for inflation (consumer prices, annual %)
inflation_data = data[
    data["Indicator Code"] == "FP.CPI.TOTL.ZG"
].copy()

# Step 2: Keep only individual countries/economies
# Exclude regional aggregates using our valid country codes
inflation_countries = inflation_data[
    inflation_data["Country Code"].isin(valid_country_codes)
].copy()

# Step 3: Select country information and years 2020-2024
inflation_countries = inflation_countries[[
    "Country Name",
    "Country Code",
    "2020",
    "2021",
    "2022",
    "2023",
    "2024"
]]

# Step 4: Display the first 10 rows
print("Inflation data:")
display(inflation_countries.head(10))

# Step 5: Count how many countries/economies we extracted
print("Number of rows:", len(inflation_countries))
```

```python

# ============================================================
# STEP 3.3: COMBINE ALL 8 ECONOMIC INDICATORS (1980-2024)
# ============================================================

import pandas as pd

# Step 1: Convert GDP growth data into long format
# Instead of one column for each year, we create one row per country per year
historical_combined = gdp_historical.melt(
    id_vars=["Country Name", "Country Code"],
    value_vars=years,
    var_name="Year",
    value_name="gdp_growth"
)

# Convert Year from text to an integer
historical_combined["Year"] = historical_combined["Year"].astype(int)


# Step 2: Define the other 5 World Bank indicators
# Each name is linked to its World Bank indicator code
wdi_indicators = {
    "inflation": "FP.CPI.TOTL.ZG",
    "unemployment": "SL.UEM.TOTL.ZS",
    "trade": "NE.TRD.GNFS.ZS",
    "investment": "NE.GDI.TOTL.ZS",
    "population_growth": "SP.POP.GROW"
}


# Step 3: Extract and merge each World Bank indicator
for name, code in wdi_indicators.items():

    # Select rows matching the indicator and valid country codes
    indicator_data = data[
        (data["Indicator Code"] == code) &
        (data["Country Code"].isin(valid_country_codes))
    ].copy()

    # Keep country codes and historical years
    indicator_data = indicator_data[
        ["Country Code"] + years
    ]

    # Convert the indicator into long format
    indicator_data = indicator_data.melt(
        id_vars="Country Code",
        value_vars=years,
        var_name="Year",
        value_name=name
    )

    # Convert Year to an integer
    indicator_data["Year"] = indicator_data["Year"].astype(int)

    # Merge using Country Code and Year
    # Keep all existing country-year rows
    historical_combined = pd.merge(
        historical_combined,
        indicator_data,
        on=["Country Code", "Year"],
        how="left",
        validate="one_to_one"
    )

    print(name, "merged successfully!")


# Step 4: Define the 2 IMF indicators
imf_indicators = {
    "government_debt": "GGXWDG_NGDP",
    "current_account": "BCA_NGDPD"
}

# IMF and World Bank use different codes for these economies
country_code_mapping = {
    "KOS": "XKX",
    "WBG": "PSE"
}


# Step 5: Extract and merge each IMF indicator
for name, code in imf_indicators.items():

    # Select the correct IMF indicator
    indicator_data = imf_data[
        imf_data["SERIES_CODE"].str.endswith(
            f".{code}.A", na=False
        )
    ].copy()

    # Extract country code from IMF series code
    # Example: SLE.GGXWDG_NGDP.A becomes SLE
    indicator_data["Country Code"] = (
        indicator_data["SERIES_CODE"].str.split(".").str[0]
    )

    # Correct country codes that differ between IMF and World Bank
    indicator_data["Country Code"] = (
        indicator_data["Country Code"].replace(country_code_mapping)
    )

    # Keep only countries found in our World Bank dataset
    indicator_data = indicator_data[
        indicator_data["Country Code"].isin(valid_country_codes)
    ].copy()

    # Select country codes and years
    indicator_data = indicator_data[
        ["Country Code"] + years
    ]

    # Convert the IMF data into long format
    indicator_data = indicator_data.melt(
        id_vars="Country Code",
        value_vars=years,
        var_name="Year",
        value_name=name
    )

    # Convert Year to an integer
    indicator_data["Year"] = indicator_data["Year"].astype(int)

    # Merge IMF data with the World Bank dataset
    historical_combined = pd.merge(
        historical_combined,
        indicator_data,
        on=["Country Code", "Year"],
        how="left",
        validate="one_to_one"
    )

    print(name, "merged successfully!")


# Step 6: Convert economic indicator values to numbers
# Any non-numeric or missing values become NaN
economic_columns = (
    ["gdp_growth"] +
    list(wdi_indicators.keys()) +
    list(imf_indicators.keys())
)

for column in economic_columns:
    historical_combined[column] = pd.to_numeric(
        historical_combined[column],
        errors="coerce"
    )


# Step 7: Sort the dataset by country and year
historical_combined = historical_combined.sort_values(
    by=["Country Code", "Year"]
).reset_index(drop=True)


# Step 8: Display the combined dataset
print("\nCombined historical economic dataset:")
display(historical_combined.head(10))


# Step 9: Check the dataset dimensions
print("Number of rows:", len(historical_combined))
print("Number of columns:", len(historical_combined.columns))
print("Number of countries:", historical_combined["Country Code"].nunique())
print("Year range:",
      historical_combined["Year"].min(),
      "-",
      historical_combined["Year"].max())
```

```python

# ==================================================
# STEP 3.4: CHECK MISSING VALUES
# ==================================================

# Step 1: Create a list of our 8 economic indicators
economic_columns = [
    "gdp_growth",
    "inflation",
    "unemployment",
    "trade",
    "investment",
    "population_growth",
    "government_debt",
    "current_account"
]

# Step 2: Count the missing values for each indicator
# isna() identifies missing values (NaN)
# sum() counts how many values are missing
missing_count = historical_combined[economic_columns].isna().sum()

# Step 3: Calculate the percentage of missing values
# mean() gives the proportion of missing values
# Multiplying by 100 converts it into a percentage
missing_percentage = (
    historical_combined[economic_columns].isna().mean() * 100
).round(2)

# Step 4: Create a summary table
missing_summary = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage (%)": missing_percentage
})

# Step 5: Sort indicators from most missing to least missing
missing_summary = missing_summary.sort_values(
    by="Missing Percentage (%)",
    ascending=False
)

# Step 6: Display the results
print("Missing values in our historical dataset:")
display(missing_summary)
```

```python

# ==================================================
# STEP 3.5: CHECK MISSING VALUES BY YEAR
# ==================================================

# Step 1: Group the data by year
# Calculate the percentage of missing values for
# government debt and unemployment in each year
missing_by_year = historical_combined.groupby("Year")[
    ["government_debt", "unemployment"]
].apply(
    lambda group: group.isna().mean() * 100
)

# Step 2: Round the percentages to 2 decimal places
missing_by_year = missing_by_year.round(2)

# Step 3: Select some years to compare
# We want to see whether recent years have fewer missing values
selected_years = [1980, 1990, 2000, 2010, 2017, 2020, 2024]

# Step 4: Display missing percentages for these years
print("Missing values by year (%):")
display(missing_by_year.loc[selected_years])
```

```python

# ==================================================
# STEP 3.6: CHECK DATA QUALITY FOR 2000-2024
# ==================================================

# Step 1: Select observations from 2000 to 2024
# between() includes both 2000 and 2024
data_2000_2024 = historical_combined[
    historical_combined["Year"].between(2000, 2024)
].copy()

# Step 2: Calculate missing percentages for all 8 indicators
# isna() checks missing values
# mean() calculates the missing proportion
# Multiplying by 100 converts it to a percentage
missing_recent = (
    data_2000_2024[economic_columns]
    .isna()
    .mean() * 100
).round(2)

# Step 3: Sort from highest to lowest missing percentage
missing_recent = missing_recent.sort_values(ascending=False)

# Step 4: Display the results
print("Missing data from 2000 to 2024 (%):")
display(missing_recent.to_frame(name="Missing Percentage (%)"))

# Step 5: Display the size of our selected dataset
print("Number of rows:", len(data_2000_2024))
print("Number of countries:",
      data_2000_2024["Country Code"].nunique())
```

```python

# ==================================================
# STEP 3.7: CHECK COMPLETE COUNTRY-YEAR OBSERVATIONS
# ==================================================

# Step 1: Check which rows have values for all 8 indicators
# notna() checks for available (non-missing) values
# all(axis=1) checks every indicator in each row
complete_rows = data_2000_2024[economic_columns].notna().all(axis=1)

# Step 2: Count how many rows have all 8 indicators
complete_count = complete_rows.sum()

# Step 3: Calculate the percentage of complete rows
complete_percentage = (
    complete_rows.mean() * 100
)

# Step 4: Count the countries with at least one complete row
complete_countries = data_2000_2024.loc[
    complete_rows, "Country Code"
].nunique()

# Step 5: Display the results
print("Total country-year rows:", len(data_2000_2024))
print("Complete rows:", complete_count)
print("Complete percentage:", round(complete_percentage, 2), "%")
print("Countries with at least one complete row:", complete_countries)
```

```python

# ==================================================
# STEP 3.8: CHECK COMPLETE OBSERVATIONS BY YEAR
# ==================================================

# Step 1: Create a copy of our dataset
yearly_data = data_2000_2024.copy()

# Step 2: Check whether each row has all 8 indicators
# True means all indicators are available
# False means at least one indicator is missing
yearly_data["complete"] = (
    yearly_data[economic_columns].notna().all(axis=1)
)

# Step 3: Group observations by year
# sum() counts complete rows because True counts as 1
# mean() calculates the proportion of complete rows
yearly_summary = yearly_data.groupby("Year")["complete"].agg(
    Complete_Rows="sum",
    Complete_Percentage="mean"
)

# Step 4: Convert proportions into percentages
yearly_summary["Complete_Percentage"] = (
    yearly_summary["Complete_Percentage"] * 100
).round(2)

# Step 5: Display selected years
selected_years = [
    2000, 2005, 2010, 2017,
    2019, 2020, 2021, 2022, 2023, 2024
]

print("Complete observations by year:")
display(yearly_summary.loc[selected_years])
```

```python

# ==================================================
# STEP 3.9: CREATE NEXT-YEAR GDP GROWTH TARGET
# ==================================================

# Step 1: Create a copy of our 2000-2024 dataset
forecast_data = data_2000_2024.copy()

# Step 2: Sort observations by country and year
# This ensures that years are in the correct order
forecast_data = forecast_data.sort_values(
    by=["Country Code", "Year"]
)

# Step 3: Create the next-year GDP growth column
# groupby() separates the data by country
# shift(-1) takes GDP growth from the following year
forecast_data["next_year_gdp_growth"] = (
    forecast_data.groupby("Country Code")["gdp_growth"]
    .shift(-1)
)

# Step 4: Keep input years 2000-2023
# We exclude 2024 as an input year because
# our dataset does not contain 2025 GDP growth
forecast_data = forecast_data[
    forecast_data["Year"] <= 2023
].copy()

# Step 5: Display Ireland's recent GDP growth
# This allows us to check whether the target is aligned
ireland_example = forecast_data[
    (forecast_data["Country Code"] == "IRL") &
    (forecast_data["Year"].between(2020, 2023))
]

print("Ireland: GDP growth and next-year target")
display(ireland_example[[
    "Country Name",
    "Year",
    "gdp_growth",
    "next_year_gdp_growth"
]])

# Step 6: Check the number of observations
print("Total rows:", len(forecast_data))

# Count observations with an available prediction target
print(
    "Rows with next-year GDP growth:",
    forecast_data["next_year_gdp_growth"].notna().sum()
)
```

```python

# ==================================================
# STEP 3.10: IDENTIFY TEST YEARS (2021-2024)
# ==================================================

# Step 1: Create a new column for the target year
# For example, input year 2023 predicts GDP growth in 2024
forecast_data["target_year"] = forecast_data["Year"] + 1

# Step 2: Select rows where the target year is 2021-2024
# These are the years we want to evaluate our models on
test_data = forecast_data[
    forecast_data["target_year"].between(2021, 2024)
].copy()

# Step 3: Count available GDP growth targets for each year
# size() counts all rows, including missing targets
# count() counts only rows with available targets
test_summary = test_data.groupby("target_year")[
    "next_year_gdp_growth"
].agg(
    total_rows="size",
    available_targets="count"
)

# Step 4: Display the results
print("Test data availability:")
display(test_summary)
```

```python

# ==================================================
# STEP 3.11: CHECK COMPLETE TEST OBSERVATIONS
# ==================================================

# Step 1: Check whether all 8 economic indicators are available
# True means no economic indicators are missing in that row
test_data["features_available"] = (
    test_data[economic_columns].notna().all(axis=1)
)

# Step 2: Check whether next-year GDP growth is available
# This is the value our ML models will try to predict
test_data["target_available"] = (
    test_data["next_year_gdp_growth"].notna()
)

# Step 3: Identify observations with both:
# All 8 indicators AND next-year GDP growth
# & means both conditions must be True
test_data["complete_observation"] = (
    test_data["features_available"] &
    test_data["target_available"]
)

# Step 4: Count available observations for each target year
test_completeness = test_data.groupby("target_year").agg(
    total_rows=("Country Code", "size"),
    available_targets=("target_available", "sum"),
    complete_observations=("complete_observation", "sum")
)

# Step 5: Display the summary
print("Test dataset completeness:")
display(test_completeness)
```

```python

# ==================================================
# STEP 3.12: CHECK TRAINING DATA AVAILABILITY
# ==================================================

# Step 1: Select observations with target years 2001-2017
# For example, 2016 indicators predict GDP growth in 2017
train_data = forecast_data[
    forecast_data["target_year"].between(2001, 2017)
].copy()

# Step 2: Check whether all 8 indicators are available
# all(axis=1) checks across the columns of each row
features_available = (
    train_data[economic_columns].notna().all(axis=1)
)

# Step 3: Check whether next-year GDP growth is available
target_available = (
    train_data["next_year_gdp_growth"].notna()
)

# Step 4: Count rows with all features and the target available
# & means both conditions must be True
complete_training_rows = (
    features_available & target_available
).sum()

# Step 5: Display the training data summary
print("Training data summary:")
print("Total rows:", len(train_data))
print("Rows with GDP targets:", target_available.sum())
print("Complete training rows:", complete_training_rows)

# Step 6: Count countries with at least one complete training row
complete_countries = train_data.loc[
    features_available & target_available,
    "Country Code"
].nunique()

print("Countries with complete training data:", complete_countries)
```

```python

# ==================================================
# STEP 3.13: CHECK VALIDATION DATA AVAILABILITY
# ==================================================

# Step 1: Select observations with target years 2018-2019
# 2017 indicators predict 2018 GDP growth
# 2018 indicators predict 2019 GDP growth
validation_data = forecast_data[
    forecast_data["target_year"].between(2018, 2019)
].copy()

# Step 2: Check whether all 8 economic indicators are available
# notna() checks for non-missing values
# all(axis=1) checks every indicator in each row
features_available = (
    validation_data[economic_columns].notna().all(axis=1)
)

# Step 3: Check whether next-year GDP growth is available
target_available = (
    validation_data["next_year_gdp_growth"].notna()
)

# Step 4: Identify rows with both features and target available
complete_validation = (
    features_available & target_available
)

# Step 5: Display the validation data summary
print("Validation data summary:")
print("Total rows:", len(validation_data))
print("Rows with GDP targets:", target_available.sum())
print("Complete validation rows:", complete_validation.sum())

# Step 6: Count countries with complete validation observations
complete_countries = validation_data.loc[
    complete_validation, "Country Code"
].nunique()

print("Countries with complete validation data:", complete_countries)
```

```python

# ==================================================
# STEP 3.14: CHECK COVID-19 YEAR DATA (2020)
# ==================================================

# Step 1: Select observations where the target year is 2020
# This means 2019 indicators are used to predict 2020 GDP growth
covid_data = forecast_data[
    forecast_data["target_year"] == 2020
].copy()

# Step 2: Check whether all 8 economic indicators are available
features_available = (
    covid_data[economic_columns].notna().all(axis=1)
)

# Step 3: Check whether the 2020 GDP growth target is available
target_available = (
    covid_data["next_year_gdp_growth"].notna()
)

# Step 4: Identify rows containing all features and the target
complete_covid = features_available & target_available

# Step 5: Display the results
print("COVID-19 stress-test data summary:")
print("Total rows:", len(covid_data))
print("Rows with GDP targets:", target_available.sum())
print("Complete observations:", complete_covid.sum())

# Step 6: Count countries with complete observations
print(
    "Countries with complete data:",
    covid_data.loc[complete_covid, "Country Code"].nunique()
)
```

```python

# ==================================================
# STEP 3.15: SAVE TRAINING, VALIDATION AND TEST DATA
# ==================================================

# Step 1: Select the columns we want to save
# Include country information, input year, target year,
# all 8 economic indicators and next-year GDP growth
columns_to_keep = [
    "Country Name",
    "Country Code",
    "Year",
    "target_year"
] + economic_columns + ["next_year_gdp_growth"]


# Step 2: Define the four datasets and their target years
# Each pair of numbers represents the start and end year
dataset_periods = {
    "GDP_Training_2001_2017.csv": (2001, 2017),
    "GDP_Validation_2018_2019.csv": (2018, 2019),
    "GDP_COVID_2020.csv": (2020, 2020),
    "GDP_Testing_2021_2024.csv": (2021, 2024)
}


# Step 3: Create and save each dataset
for filename, (start_year, end_year) in dataset_periods.items():

    # Select rows based on the year being predicted
    selected_data = forecast_data[
        forecast_data["target_year"].between(
            start_year, end_year
        )
    ].copy()

    # Keep only the columns we need
    selected_data = selected_data[columns_to_keep]

    # Save as CSV without the pandas row index
    selected_data.to_csv(filename, index=False)

    # Display confirmation
    print(filename, "- saved successfully!")
    print("Rows:", len(selected_data))
    print()


# Step 4: Save the complete forecasting dataset
# This is useful as a master copy for future analysis
forecast_data[columns_to_keep].to_csv(
    "GDP_Full_Forecasting_Data_DRAFT.csv",
    index=False
)

print("Full forecasting dataset saved!")
```
